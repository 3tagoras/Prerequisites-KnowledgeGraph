```python
from github import Github

g = Github()  # or Github() for anonymous
repo = g.get_repo('itsshivampal/Prerequisite-Learning-Dataset')
tree = repo.get_git_tree('master', recursive=True)

folders = [e.path for e in tree.tree if e.type == 'tree']
files   = [e.path for e in tree.tree if e.type == 'blob']
print(f"{len(folders)} folders, {len(files)} files")
```

```python
from github_downloader import download_from_github
from pathlib import Path

BASE_URL = "https://github.com/itsshivampal/Prerequisite-Learning-Dataset/blob/master/"

for f in files:
    sections = f.split("/")
    if len(sections)==3 and sections[0]!= 'src':
        folder = sections[0]
        #Path("../data/raw/"+folder).mkdir(parents=True, exist_ok=True)
        download_from_github(
            BASE_URL+f,
            dest_folder="../data/raw/PLD/"+folder
        )
```
